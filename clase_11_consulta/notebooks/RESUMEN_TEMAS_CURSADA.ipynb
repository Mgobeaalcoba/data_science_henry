**Regla de oro desde la Clase 01:** El `train_test_split` se realiza **antes** de calcular estadísticos de imputación (mediana, moda) o de ajustar escaladores (`StandardScaler`, `MinMaxScaler`). Ajustar transformadores sobre todo el dataset introduce **Data Leakage**.

---

### **Clase 02: Regresión Lineal Simple y Múltiple**

$$\hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \dots + \beta_p x_p$$

- **Comparación de importancia entre variables:** No deben compararse los coeficientes brutos $\beta_j$ cuando las variables están en distintas unidades (ej. `income` en miles de dólares vs. `satisfaction_score` de $1$ a $5$). Se deben usar **coeficientes estandarizados** (entrenando sobre `X_train` escalado con `StandardScaler`).
- **Métricas clave:**
  - $\text{MAE} = \frac{1}{n}\sum |y_i - \hat{y}_i|$ (en las mismas unidades del *target*).
  - $\text{RMSE} = \sqrt{\frac{1}{n}\sum (y_i - \hat{y}_i)^2}$ (penaliza errores grandes).
  - $R^2 = 1 - \frac{SS_{\text{res}}}{SS_{\text{tot}}}$ y **$R^2$ Ajustado**: $R^2_{\text{adj}} = 1 - \frac{(1-R^2)(n-1)}{n-p-1}$ (penaliza agregar variables irrelevantes).
- **Supuestos de los residuos ($e_i = y_i - \hat{y}_i$):** media cero, **homocedasticidad** (varianza constante frente a $\hat{y}_i$) y **normalidad**.

---

### **Clase 03: Regresión Logística (Avance 1 — Proyecto Integrador FinanceGuard)**

$$\sigma(z) = \frac{1}{1 + e^{-z}}, \quad z = \beta_0 + \sum_{j=1}^p \beta_j x_j, \quad \text{Odds Ratio} = e^{\beta_j}$$

- En problemas desbalanceados (como `Churn_Modelling.csv`, con $20,37\%$ de *churn*), un modelo sin balanceo tiene alta *Accuracy* pero bajo *Recall*.
- Usar `class_weight="balanced"` (o ajustar el umbral de decisión) eleva sustancialmente el **Recall** de la clase minoritaria.

---

### **Clase 04: KNN, Árboles de Decisión, SVM y Métricas de Negocio**

| Algoritmo | Geometría de decisión | ¿Requiere escalado? | Fortaleza principal |
|---|---|---|---|
| **KNN** | Vecindad por distancia euclidiana | **Sí** (`StandardScaler`) | Simple, no paramétrico |
| **Árbol de Decisión** | Cortes ortogonales por impureza Gini/Entropía | No | Alta interpretabilidad |
| **SVM (RBF)** | Hiperplano de máximo margen en espacio transformado | **Sí** (`StandardScaler`) | Excelente separación no lineal (`AUC`) |

- **Preprocesamiento diferenciado:** Escalar variables continuas con `StandardScaler`, mantener variables binarias ($0/1$) en `"passthrough"` y codificar nominales con `OneHotEncoder(drop="first")`.
- **Elección de métrica según costo de error:** Cuando un **Falso Negativo** (perder un *lead* que iba a convertir o un cliente que se fuga) es más costoso que un **Falso Positivo** (hacer un llamado comercial extra), priorizamos **Recall**, **F1-score** y **ROC-AUC**, y ajustamos el umbral de probabilidad por debajo de $0,50$.

## 🔧 3. MÓDULO 2: MODELOS DE ENSAMBLE Y OPTIMIZACIÓN (Clases 05–06)

### **Clase 05: Modelos de Ensamble (*Bagging* vs. *Boosting*)**

| Aspecto | Árbol Individual (`DecisionTree`) | Random Forest (*Bagging*) | XGBoost / Gradient Boosting (*Boosting*) |
|---|---|---|---|
| **Construcción** | Un único árbol profundo | Árboles independientes en paralelo (*bootstrap*) | Árboles secuenciales que corrigen residuos |
| **Efecto principal** | Bajo sesgo, **alta varianza** (*overfitting*) | **Reduce la varianza** promediando árboles | **Reduce sesgo y varianza** simultáneamente |
| **Diagnóstico clave** | $R^2_{\text{train}} \approx 1,00 \gg R^2_{\text{test}}$ | Brecha Train–Test moderada | Mejor $R^2_{\text{test}}$ y menor $\text{RMSE}$ con hiperparámetros regulados |

---

### **Clase 06: Optimización de Hiperparámetros (Avance 2 — Proyecto Integrador)**

Comparación de estrategias de búsqueda con validación cruzada estratificada (`StratifiedKFold`, métricas **F1**, **ROC-AUC** y **PR-AUC**):
1. **Grid Search (`GridSearchCV`):** evalúa exhaustivamente el producto cartesiano de una grilla discreta. Costoso cuando crece el número de hiperparámetros.
2. **Random Search (`RandomizedSearchCV`):** muestrea combinaciones aleatorias; explora mejor hiperparámetros continuos a igual presupuesto de cómputo.
3. **Búsqueda Bayesiana (`Optuna` — TPE):** aprende de los *trials* previos para concentrar la búsqueda en las regiones más prometedoras del espacio de hiperparámetros.

### **Clase 07: Segmentación con K-Means (`Mall_Customers.csv`)**
- **Estandarización obligatoria:** K-Means minimiza la inercia (suma de distancias euclidianas cuadráticas al centroide). Sin `StandardScaler`, la variable de mayor rango numérico domina la partición.
- **Selección de $k$:** Combinar el **Método del Codo (Inercia)** con el **Coeficiente de Silueta** ($k=6$ en el espacio 3D `Age`, `Annual Income`, `Spending Score`, donde el segmento de ingreso y gasto medios se divide por edad en jóvenes vs. adultos mayores).

---

### **Clase 08: Aprendizaje No Supervisado II (Avance 3 — Proyecto Integrador FinanceGuard)**
- **Higiene no supervisada:** La etiqueta `Exited` **se excluye del entrenamiento** de `KMeans`, `DBSCAN`, `PCA` y `t-SNE`, y se usa únicamente *ex-post* para cuantificar la tasa de *churn* de cada segmento descubierto.
- **K-Means ($K=3$):** Alcanza el máximo global del coeficiente de silueta (`0.1334`), identificando un segmento crítico (Cluster 0: mercado alemán de alto balance) con una tasa de *churn* del **$32,4\%$** (casi el doble que los otros dos segmentos).
- **PCA y matriz de cargas (`pca.components_`):** Permite proyectar en 2D y auditar qué variables originales impulsan cada componente principal.
- **DBSCAN y curva $k$-NN (`NearestNeighbors`):** Permite elegir fundamentadamente el radio `eps` en el codo de distancias y aislar clientes atípicos (*outliers*, etiqueta `-1`) sin forzarlos dentro de un centroide.
- **t-SNE:** Proyección no lineal que preserva vecindades locales para inspeccionar subestructuras en 2D.

### **Clase 09: Análisis y Pronóstico de Series Temporales (`CityScoot`)**

2. **Diagnóstico estadístico:**
   - **Test ADF (`adfuller`):** verifica si la serie es estacionaria ($p < 0,05$) o no estacionaria por presencia de tendencia ($p > 0,05$).
   - **Gráficos ACF y PACF:** revelan la memoria autorregresiva y los picos estacionales en múltiplos del período ($7, 14, 21, 28$ días).
3. **Forecasting supervisado vs. Baselines:**
   - Todo modelo de series temporales debe compararse contra el **Baseline de Persistencia Diaria (`lag_1`)** y el **Baseline Estacional (`lag_7`)** usando **MAE, RMSE y MAPE**, respetando estrictamente el corte cronológico (`shuffle=False`).

### **Clase 10: Deep Learning con PyTorch — Red LSTM (`econotrend_vix_sim.csv`)**
- **Preparación sin *leakage* ni pérdida de datos de prueba:**
  1. Dividir cronológicamente en `train` ($80\%$) y `test` ($20\%$) **antes** de escalar.
  2. Ajustar `MinMaxScaler` solo en `train`.
  3. Construir las ventanas de `test` concatenando los últimos `LOOKBACK = 10` días de `train` (`make_test_sequences`) para evaluar los $261$ días completos de prueba.
- **Lección central del caso VIX:** Tanto la **LSTM** como la **Regresión Lineal** sobre los 10 rezagos explican $\sim 67\%$ de la varianza ($R^2 \approx 0,67$, $\text{MAE} \approx 0,93\text{ pts}$), superando al *baseline* de persistencia. Cuando la dinámica es predominantemente lineal, un modelo simple iguala a una red de $50.497$ parámetros (**Principio de Parsimonia**).

### Plantilla 1: Pipeline Completo de Clasificación / Regresión Tabular

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import classification_report, roc_auc_score
from xgboost import XGBClassifier

# 1. Separar X e y ANTES de cualquier transformación (descartando IDs)
X = df.drop(columns=["id_cliente", "target"])
y = df["target"]

# 2. Split estratificado PRIMERO
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Preprocesador por tipo de columna (imputación + escalado + encoding dentro del Pipeline)
num_cols = ["edad", "ingresos", "visitas"]
bin_cols = ["tiene_tarjeta"]
cat_cols = ["region"]

num_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("num", num_pipe, num_cols),
    ("bin", "passthrough", bin_cols),
    ("cat", cat_pipe, cat_cols),
])

# 4. Pipeline completo (garantiza 0 Data Leakage en Cross-Validation y en Test)
clf = Pipeline([
    ("prep", preprocessor),
    ("model", XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05, eval_metric="logloss", random_state=42)),
])

# 5. Validación cruzada SOLO sobre train
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_auc = cross_val_score(clf, X_train, y_train, cv=cv, scoring="roc_auc")
print(f"ROC-AUC CV (5 folds): {cv_auc.mean():.4f} +/- {cv_auc.std():.4f}")

# 6. Ajuste final en train y evaluación en test
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
y_proba = clf.predict_proba(X_test)[:, 1]
print(classification_report(y_test, y_pred))
print(f"ROC-AUC Test: {roc_auc_score(y_test, y_proba):.4f}")
```

### Plantilla 2: Serie Temporal con Red LSTM en PyTorch (Sin *Data Leakage*)

```python
import numpy as np
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error

# 1. Split cronológico ANTES de escalar
LOOKBACK = 10
split_idx = int(len(serie_valores) * 0.80)
train_raw = serie_valores[:split_idx].reshape(-1, 1)
test_raw  = serie_valores[split_idx:].reshape(-1, 1)

# 2. Ajustar MinMaxScaler SOLO en train
scaler = MinMaxScaler()
train_sc = scaler.fit_transform(train_raw)
test_sc  = scaler.transform(test_raw)

# 3. Ventanas deslizantes preservando el 100% de los días de test
def make_sequences(data, lookback=10):
    X, y = [], []
    for i in range(lookback, len(data)):
        X.append(data[i - lookback:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

X_train, y_train = make_sequences(train_sc, LOOKBACK)
X_test, y_test   = make_sequences(np.vstack([train_sc[-LOOKBACK:], test_sc]), LOOKBACK)

X_train_t = torch.FloatTensor(X_train).unsqueeze(-1)
y_train_t = torch.FloatTensor(y_train).unsqueeze(-1)
X_test_t  = torch.FloatTensor(X_test).unsqueeze(-1)

# 4. Definir y entrenar LSTM con las 5 líneas esenciales de PyTorch
class LSTMRegressor(nn.Module):
    def __init__(self, hidden_size=64):
        super().__init__()
        self.lstm = nn.LSTM(input_size=1, hidden_size=hidden_size, num_layers=2, batch_first=True, dropout=0.2)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        return self.fc(out[:, -1, :])

model = LSTMRegressor()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(50):
    model.train()
    optimizer.zero_grad()                  # 1. Limpiar gradientes
    pred = model(X_train_t)                # 2. Forward pass
    loss = criterion(pred, y_train_t)      # 3. Calcular pérdida
    loss.backward()                        # 4. Backward pass
    optimizer.step()                       # 5. Actualizar pesos

# 5. Evaluar en escala original desnormalizando con inverse_transform
model.eval()
with torch.no_grad():
    pred_sc = model(X_test_t).numpy()
y_real = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()
y_pred = scaler.inverse_transform(pred_sc).flatten()
print(f"MAE en unidades reales: {mean_absolute_error(y_real, y_pred):.4f}")
```

## 🛠️ 7. CAJA DE HERRAMIENTAS EJECUTABLE PARA EL PROYECTO FINAL

A continuación definimos y probamos dos funciones listas para copiar y usar en cualquier proyecto de **Clasificación** o **Regresión**. Ambas calculan simultáneamente las métricas en `Train` y `Test` para diagnosticar **sobreajuste (*overfitting*)** de un vistazo.

````python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification, make_regression
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, mean_absolute_error, mean_squared_error, r2_score
)

def evaluar_clasificador(modelo, X_train, y_train, X_test, y_test, nombre="Modelo"):
    modelo.fit(X_train, y_train)
    pred_tr = modelo.predict(X_train)
    pred_te = modelo.predict(X_test)
    proba_te = modelo.predict_proba(X_test)[:, 1] if hasattr(modelo, "predict_proba") else None

    f1_tr = f1_score(y_train, pred_tr)
    f1_te = f1_score(y_test, pred_te)
    return {
        "Modelo": nombre,
        "Accuracy_Test": accuracy_score(y_test, pred_te),
        "Precision_Test": precision_score(y_test, pred_te, zero_division=0),
        "Recall_Test": recall_score(y_test, pred_te, zero_division=0),
        "F1_Train": f1_tr,
        "F1_Test": f1_te,
        "Brecha_F1 (Overfit)": f1_tr - f1_te,
        "ROC_AUC_Test": roc_auc_score(y_test, proba_te) if proba_te is not None else np.nan,
    }

def evaluar_regresor(modelo, X_train, y_train, X_test, y_test, nombre="Modelo"):
    modelo.fit(X_train, y_train)
    pred_tr = modelo.predict(X_train)
    pred_te = modelo.predict(X_test)
    r2_tr = r2_score(y_train, pred_tr)
    r2_te = r2_score(y_test, pred_te)
    return {
        "Modelo": nombre,
        "MAE_Test": mean_absolute_error(y_test, pred_te),
        "RMSE_Test": np.sqrt(mean_squared_error(y_test, pred_te)),
        "R2_Train": r2_tr,
        "R2_Test": r2_te,
        "Brecha_R2 (Overfit)": r2_tr - r2_te,
    }

# Demostración rápida sobre un problema sintético de clasificación
X_demo, y_demo = make_classification(n_samples=600, n_features=10, n_informative=6, weights=[0.75, 0.25], random_state=42)
X_tr_d, X_te_d, y_tr_d, y_te_d = train_test_split(X_demo, y_demo, test_size=0.25, stratify=y_demo, random_state=42)

comparativa_demo = pd.DataFrame([
    evaluar_clasificador(LogisticRegression(class_weight="balanced"), X_tr_d, y_tr_d, X_te_d, y_te_d, "Baseline Logística Balanceada"),
    evaluar_clasificador(RandomForestClassifier(max_depth=5, random_state=42), X_tr_d, y_tr_d, X_te_d, y_te_d, "Random Forest (max_depth=5)"),
]).set_index("Modelo").round(3)

print("Demostración ejecutable de la función evaluar_clasificador:")
comparativa_demo
````

## 🎯 8. PROYECTO FINAL: GUÍA PASO A PASO Y ERRORES COMUNES

### Matriz de Decisión Antes de Escribir Código

| Pregunta | Decisión a justificar en el notebook |
|---|---|
| ¿El *target* es continuo, binario, multiclase o inexistente? | Define la familia de algoritmos y el tipo de validación (`StratifiedKFold` vs. `KFold` vs. `TimeSeriesSplit`). |
| ¿Cuál es el *baseline* obligatorio? | Media (`DummyRegressor`) en regresión, clase mayoritaria/regresión logística simple en clasificación, o persistencia (`lag_1`/`lag_7`) en series temporales. |
| ¿Cómo se previene el *Data Leakage*? | Dividir `train`/`test` **antes** de imputar, codificar o escalar, y encapsular transformaciones en `Pipeline` + `ColumnTransformer`. |
| ¿Qué métrica representa el costo de negocio? | `MAE`/`RMSE` en dólares o unidades físicas; `Recall`/`F1`/`PR-AUC` cuando los Falsos Negativos son costosos; `Silhouette` + perfil de negocio en clustering. |
| ¿Cómo se audita el sobreajuste? | Comparar siempre la métrica en `Train` vs. `Test` (o `CV`) antes de elegir el modelo ganador. |

---

### Los 5 Errores Más Comunes y Cómo Evitarlos

1. **Hacer `.fit_transform()` sobre todo el dataset antes de `train_test_split`:** Contamina el entrenamiento con estadísticos del conjunto de prueba (*Data Leakage*).
2. **Hacer `shuffle=True` en series temporales:** Mezcla pasado y futuro. En series de tiempo el corte debe ser estrictamente cronológico.
3. **Confiar solo en *Accuracy* con clases desbalanceadas:** En un problema con $10\%$ de positivos, predecir siempre `0` da $90\%$ de *Accuracy* y $0\%$ de *Recall*.
4. **No comparar contra un modelo *baseline*:** Un $\text{MAE}$ aislado no dice nada si no demostramos que mejora al predictor ingenuo.
5. **Comparar coeficientes lineales sin estandarizar o escalar variables binarias sin criterio:** Para comparar importancia en modelos lineales hay que usar variables estandarizadas, y en `ColumnTransformer` conviene dejar las variables binarias ($0/1$) en `"passthrough"`.
